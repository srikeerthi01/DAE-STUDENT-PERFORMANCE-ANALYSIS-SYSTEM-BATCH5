```python
#calculate total marks
import pandas as pd
marks = data[
    ['math_score',
    'english_score',
    'science_score']
].to_numpy()
total = np.sum(marks,axis=1)
print(total)
```

```python
#calulate average marks
import numpy as np
average = np.mean(marks,axis=1)
print(average)
```

```python
#average performance by extracurricular activity
import pandas as pd
import numpy as np
result = data.groupby('extra_activities')['math_score'].mean()
print(result)
```

```python
#performance category
import pandas as pd
import numpy as np
data['performance'] = np.where(
    data['overall_score'] >= 80,
    'Excellent',
    'Needs Improvement'
)
print(data[['student_id','performance']])
#this is to find out how mnay students needs improve their performance
print(data['performance'].value_counts())
```

```python
# Adding calculated columns
#finding total marks,average marks of the students

data['total_marks'] = (
    data['math_score'] +
    data['science_score'] +
    data['english_score']
)

data['average_marks'] = data['total_marks'] / 3

print(data[['student_id', 'total_marks', 'average_marks']].head(10))
```

```python
# Adding grade based on average marks

data['grade'] = np.where(data['average_marks'] >= 90, 'A',
                 np.where(data['average_marks'] >= 80, 'B',
                 np.where(data['average_marks'] >= 70, 'C',
                 np.where(data['average_marks'] >= 60, 'D',
                 np.where(data['average_marks'] >= 50, 'E', 'F')))))

print(data[['student_id', 'average_marks', 'grade']].head(10))
```

```python
# Grade distribution
#simply counting the count of students who got which grade

grade_count = data['grade'].value_counts()

print(grade_count)
```

```python
# Subject-wise average marks
#based on the students subject wise marks, each subject has their own averages
#this helps to find which subject students are not focusing more

subject_averages = data[['math_score',
                         'science_score',
                         'english_score']].mean()

print(subject_averages)
```

```python
#this is the subject wise graph for each subject
#everything is somehow in the same range because it got  similar averages
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 5))

plt.bar(subject_averages.index, subject_averages.values)

plt.xlabel("Subjects")
plt.ylabel("Average Marks")
plt.title("Subject-wise Average Marks")

plt.show()
```

```python
# Pie chart for distributing grades
#to know about the grades of the students picotiral

plt.figure(figsize=(7, 7))

plt.pie(grade_count.values,
        labels=grade_count.index,
        autopct='%1.1f%%')

plt.title("Grade Distribution")

plt.show()
```

```python
# to identify the person who scored high among all the students
#(the current person having highest average of all)
topper = data.loc[data['average_marks'].idxmax()]

print("Topper Details:")
print(topper[['student_id',
              'math_score',
              'science_score',
              'english_score',
              'total_marks',
              'average_marks',
              'grade']])
```

```python
# Topper prediction
#this helps us to identify the students who performs well and
#they can be trained well to get more score for the finals also
#students with average abouve 90 are considered
#We use a rule-based prediction approach where students with an average mark of 90 or above are identified as potential topper candidates."

predicted_toppers = data[data['average_marks'] >= 90]

print("Potential Topper Candidates:")
print(predicted_toppers[['student_id',
                         'average_marks',
                         'grade']])
```

```python
# Student Performance Report
#this is the final report we can submit to the authorities for taking better acedamic decisions

print("===== STUDENT PERFORMANCE REPORT =====")

print("\nTotal Number of Students:", len(data))

print("\nGrade Distribution:")
print(grade_count)

print("\nPerformance Distribution:")
print(data['performance'].value_counts())

print("\nSubject-wise Average Marks:")
print(subject_averages)

print("\nTopper Details:")
print("Student ID:", topper['student_id'])
print("Average Marks:", round(topper['average_marks'], 2))
print("Grade:", topper['grade'])

print("\nPotential Topper Candidates:",
      len(predicted_toppers))
```
